```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Load the three datasets
export_df = pd.read_csv("Coffee_export-1.csv")
import_df = pd.read_csv("Coffee_import-1.csv")
production_df = pd.read_csv("Coffee_production-1.csv")

# Display the first five rows of each dataset
print("COFFEE EXPORT DATA")
display(export_df.head())

print("COFFEE IMPORT DATA")
display(import_df.head())

print("COFFEE PRODUCTION DATA")
display(production_df.head())
```

```python
datasets = {
    "Coffee Exports": export_df,
    "Coffee Imports": import_df,
    "Coffee Production": production_df
}

for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(name)
    print("=" * 50)

    print("\nDataset dimensions (rows, columns):")
    print(df.shape)

    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nData types and non-null counts:")
    df.info()

    print("\nMissing values per column:")
    print(df.isnull().sum())

    print("\nDuplicate rows:")
    print(df.duplicated().sum())

    print("\nUnique values per column:")
    for column in df.columns:
        print(f"{column}: {df[column].nunique()} unique values")
```

```python
def find_columns(df, keywords):
    return [
        col for col in df.columns
        if any(word in str(col).lower() for word in keywords)
    ]

for name, df in datasets.items():
    print(f"\n{name}")

    year_columns = find_columns(
        df, ["year", "date", "time"]
    )
    country_columns = find_columns(
        df, ["country", "nation", "reporter", "partner"]
    )

    if year_columns:
        for col in year_columns:
            values = pd.to_numeric(df[col], errors="coerce").dropna()
            if not values.empty:
                print(
                    f"{col}: {int(values.min())}–{int(values.max())}"
                )
            else:
                print(f"{col}: inspect date formatting")

    else:
        print("No obvious year column detected.")

    if country_columns:
        for col in country_columns:
            print(f"\nCountries or locations in '{col}':")
            print(sorted(df[col].dropna().astype(str).unique()))
    else:
        print("No obvious country column detected.")
```

```python
for name, df in datasets.items():
    print(f"\n{name} — Descriptive Statistics")
    display(df.describe())

    numeric_df = df.select_dtypes(include="number")

    if not numeric_df.empty:
        summary = pd.DataFrame({
            "Minimum": numeric_df.min(),
            "Maximum": numeric_df.max(),
            "Average": numeric_df.mean()
        })

        display(summary)
```

```python
for name, df in datasets.items():
    print(f"\n{name} — Data Quality Assessment")

    # Missing values
    missing = df.isnull().sum()
    print("\nMissing values:")
    print(missing[missing > 0] if missing.any() else "None found")

    # Duplicate rows
    duplicate_count = df.duplicated().sum()
    print(f"\nDuplicate rows: {duplicate_count}")

    # Whitespace in text fields
    text_columns = df.select_dtypes(include=["object", "string"]).columns

    print("\nText fields with leading or trailing spaces:")
    found_spaces = False

    for col in text_columns:
        values = df[col].dropna().astype(str)
        if values.str.strip().ne(values).any():
            print(col)
            found_spaces = True

    if not found_spaces:
        print("None found")

    # Negative values in numeric columns
    numeric_df = df.select_dtypes(include="number")
    negative_counts = (numeric_df < 0).sum()
    negative_counts = negative_counts[negative_counts > 0]

    print("\nNegative numerical values:")
    print(negative_counts if not negative_counts.empty else "None found")
```
